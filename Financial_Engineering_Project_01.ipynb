## 6. Methodology

### 6.1 Asset Volatility

Volatility was measured using the standard deviation of daily asset returns and annualised using 252 trading days.

Annualised volatility provides an estimate of the magnitude of an asset's return fluctuations over a year.

### 6.2 Correlation Analysis

Pearson correlation coefficients were calculated to examine the relationships between the daily returns of the five assets.

The correlation matrix was used to identify assets with similar or different return behaviour and assess potential diversification benefits.

### 6.3 Equal-Weight Portfolio Construction

An equal-weight portfolio was constructed by allocating 20% of the investment to each of the five assets.

The portfolio's daily return was calculated as the weighted sum of individual asset returns.

The portfolio was assumed to be rebalanced to its target weights daily. Transaction costs and taxes were excluded.

### 6.4 Value at Risk (VaR)

Value at Risk estimates a loss threshold over a specified time horizon at a given confidence level.

Two approaches were applied:

**Historical VaR:** Uses the empirical distribution of observed portfolio returns to estimate the loss threshold.

**Parametric VaR:** Assumes portfolio returns follow a normal distribution and uses the estimated mean, standard deviation and normal quantile to calculate the loss threshold.

Both approaches were initially calculated at 95% and 99% confidence levels using a one-day horizon.

### 6.5 Expected Shortfall (ES)

Expected Shortfall estimates the average loss among observations that exceed the VaR loss threshold.

It provides additional information about the severity of losses in the tail of the return distribution.

Historical Expected Shortfall was calculated using the average of portfolio losses beyond the corresponding historical VaR threshold.

### 6.6 Monte Carlo Simulation

Monte Carlo simulation was used to generate 20,000 hypothetical ten-day portfolio return paths.

Daily portfolio returns were randomly sampled with replacement from the historical return series. The sampled returns were compounded to estimate the distribution of possible ten-day portfolio values.

Ten-day VaR and Expected Shortfall were then calculated at 95% and 99% confidence levels.

### 6.7 Portfolio Performance Measurement

Portfolio performance was evaluated using:

- **Total compounded return:** Measures cumulative portfolio growth over the observation period.
- **Compound Annual Growth Rate (CAGR):** Measures the equivalent annual compounded growth rate.
- **Maximum drawdown:** Measures the largest percentage decline from a previous portfolio peak.
- **Sharpe ratio:** Measures annualised excess return relative to annualised volatility. A zero risk-free rate was assumed.

## 10. Out-of-Sample Value at Risk Backtesting

### 10.1 Backtesting Procedure

To evaluate the performance of the VaR models, the return dataset was divided chronologically into two subsets:

- **Training sample:** 1,054 daily observations (70%).
- **Testing sample:** 453 daily observations (30%).

Historical and parametric VaR thresholds were estimated using the training sample and subsequently evaluated against returns in the testing sample.

A VaR exception occurs when the actual portfolio loss exceeds the estimated VaR threshold.

The observed exception frequency was compared with the theoretical exception frequency associated with each confidence level.

### 10.2 VaR Exception Results

| Confidence Level | Expected Exception Rate | Historical Exceptions | Historical Rate | Parametric Exceptions | Parametric Rate |
|---|---:|---:|---:|---:|---:|
| 95% | 5.00% | 11 | 2.43% | 10 | 2.21% |
| 99% | 1.00% | 3 | 0.66% | 4 | 0.88% |

Both models recorded fewer exceptions than theoretically expected in the testing period.

At the 95% confidence level, the historical model recorded 11 exceptions, compared with an expected frequency of approximately 23 observations. The parametric model recorded 10 exceptions.

At the 99% confidence level, the historical model recorded three exceptions, while the parametric model recorded four.

These results indicate that both models produced relatively conservative VaR thresholds during this particular testing period.

### 10.3 Exact Binomial Test of VaR Unconditional Coverage

An exact binomial test was conducted to examine whether the observed exception frequency differed significantly from the expected exception probability.

The null hypothesis assumes that the probability of a VaR exception equals the theoretically expected probability.

The significance level was set at 5%.

| Confidence Level | Historical Model p-value | Parametric Model p-value |
|---|---:|---:|
| 95% | 0.0092 | 0.0047 |
| 99% | 0.6378 | 1.0000 |

At the 95% confidence level, both p-values are below 0.05. Therefore, the null hypothesis is rejected for both models. The observed exception frequencies were statistically inconsistent with the expected 5% exception probability.

Since both models recorded fewer exceptions than expected, the results suggest that their VaR thresholds were conservative during the testing period.

At the 99% confidence level, both p-values exceed 0.05. Therefore, there is insufficient statistical evidence to reject the null hypothesis at this confidence level.

However, this does not establish that either model is accurate. With only 453 testing observations, approximately 4.53 exceptions would be expected at the 99% level. The small expected number of exceptions limits the statistical power of the test.

### 10.4 Interpretation and Limitations

The backtesting results provide evidence that the two VaR approaches behaved differently across confidence levels.

At 95%, both models produced significantly fewer exceptions than expected. At 99%, the available evidence was insufficient to identify a statistically significant difference from the expected exception frequency.

The exact binomial test evaluates unconditional coverage but does not examine whether exceptions occur independently or cluster during periods of market stress.

Further analysis could incorporate exception-independence and conditional-coverage tests, such as the Christoffersen test, to provide a more comprehensive assessment of VaR model performance.

## 12. Limitations of the Study

The analysis has several limitations that should be considered when interpreting the results.

1. **Historical data limitation:** Historical market behaviour may not accurately represent future market conditions.

2. **Equal-weight allocation:** The portfolio uses fixed 20% target weights rather than an optimisation-based allocation strategy.

3. **Transaction costs and taxes:** The analysis excludes transaction costs, management fees, taxes and other implementation expenses.

4. **Monte Carlo assumptions:** The simulation uses independent sampling with replacement from historical daily returns. It does not explicitly preserve volatility clustering, serial dependence or changing market regimes.

5. **Overlapping historical windows:** The rolling ten-day historical returns overlap, meaning successive observations are not independent.

6. **Backtesting sample size:** The 453-observation testing sample provides relatively few expected exceptions at the 99% confidence level, limiting the statistical power of the test.

7. **Coverage testing:** The exact binomial test evaluates unconditional exception frequency but does not test exception independence or clustering.

These limitations mean that the results should be interpreted as historical empirical analysis rather than a guarantee of future portfolio performance or risk.

```python

# Import required libraries

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import yfinance as yf
from scipy.stats import norm

# Display settings
pd.set_option('display.float_format', '{:.4f}'.format)

print("Financial Engineering Environment Successfully Set Up!")
```

```python

# Define our financial assets
tickers = ['SPY', 'QQQ', 'GLD', 'TLT', 'EEM']

# Download historical market data
prices = yf.download(
    tickers,
    start='2020-01-01',
    end='2026-01-01',
    auto_adjust=True,
    progress=False
)['Close']

# Display the first five observations
print("Historical Asset Prices:")
display(prices.head())

# Check dataset dimensions
print("Dataset Shape:", prices.shape)
```

```python

# Check missing values
print("Missing Values:")
print(prices.isnull().sum())

# Check duplicate dates
print("\nDuplicate Dates:")
print(prices.index.duplicated().sum())

# Summary statistics
print("\nPrice Summary:")
display(prices.describe())

# Check date range
print("\nFirst Date:", prices.index.min())
print("Last Date:", prices.index.max())
```

```python

# Calculate daily percentage returns
returns = prices.pct_change().dropna()

# Display the first five daily returns
print("Daily Asset Returns:")
display(returns.head())

# Summary statistics
print("\nDaily Return Statistics:")
display(returns.describe())

# Confirm dataset dimensions
print("\nReturns Dataset Shape:", returns.shape)
```

```python

# Calculate annualised volatility
daily_volatility = returns.std()

annual_volatility = daily_volatility * np.sqrt(252)

# Display volatility results as percentages
volatility_table = pd.DataFrame({
    'Daily Volatility (%)': daily_volatility * 100,
    'Annual Volatility (%)': annual_volatility * 100
})

volatility_table = volatility_table.sort_values(
    by='Annual Volatility (%)',
    ascending=False
)

display(volatility_table)

# Visualise annual volatility
volatility_table['Annual Volatility (%)'].plot(
    kind='bar',
    figsize=(10, 5),
    title='Annualised Volatility of Five Financial Assets'
)

plt.ylabel('Volatility (%)')
plt.xlabel('Financial Assets')
plt.xticks(rotation=0)
plt.grid(axis='y', alpha=0.3)
plt.show()
```

```python

# Calculate correlation between asset returns
correlation_matrix = returns.corr()

print("Asset Return Correlation Matrix:")
display(correlation_matrix.round(3))

# Visualise the correlation matrix
plt.figure(figsize=(9, 6))

plt.imshow(
    correlation_matrix,
    cmap='coolwarm',
    vmin=-1,
    vmax=1
)

plt.colorbar(label='Correlation')

plt.xticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns
)

plt.yticks(
    range(len(correlation_matrix.index)),
    correlation_matrix.index
)

# Display correlation values
for i in range(len(correlation_matrix)):
    for j in range(len(correlation_matrix)):
        plt.text(
            j, i,
            f'{correlation_matrix.iloc[i, j]:.2f}',
            ha='center',
            va='center'
        )

plt.title('Correlation Between Financial Asset Returns')
plt.tight_layout()
plt.show()
```

```python

# Define equal portfolio weights
weights = np.array([0.20, 0.20, 0.20, 0.20, 0.20])

# Calculate daily portfolio returns
portfolio_returns = returns.dot(weights)

# Calculate annualised portfolio volatility
portfolio_volatility = portfolio_returns.std() * np.sqrt(252)

# Compare individual assets with portfolio
risk_comparison = pd.DataFrame({
    'Annual Volatility (%)': annual_volatility * 100
})

risk_comparison.loc['Equal Weight Portfolio'] = (
    portfolio_volatility * 100
)

# Display results
print("Portfolio Risk Comparison:")
display(
    risk_comparison.sort_values(
        by='Annual Volatility (%)',
        ascending=False
    )
)

# Calculate annualised portfolio return
annual_return = portfolio_returns.mean() * 252

print("\nPortfolio Annualised Average Return:",
      round(annual_return * 100, 2), "%")

print("Portfolio Annualised Volatility:",
      round(portfolio_volatility * 100, 2), "%")
```

```python

# Investment amount
investment = 10000

# Confidence levels
confidence_levels = [0.95, 0.99]

# Store risk results
risk_results = []

for confidence in confidence_levels:

    # Calculate lower-tail return threshold
    percentile = 1 - confidence

    historical_threshold = portfolio_returns.quantile(percentile)

    # Historical VaR
    historical_var = -historical_threshold * investment

    # Historical Expected Shortfall
    tail_returns = portfolio_returns[
        portfolio_returns <= historical_threshold
    ]

    historical_es = -tail_returns.mean() * investment

    # Parametric VaR using normal distribution
    mean_return = portfolio_returns.mean()
    daily_std = portfolio_returns.std()

    z_score = norm.ppf(percentile)

    parametric_var = -(
        mean_return + z_score * daily_std
    ) * investment

    risk_results.append({
        'Confidence Level': f'{confidence:.0%}',
        'Historical VaR ($)': historical_var,
        'Historical ES ($)': historical_es,
        'Parametric VaR ($)': parametric_var
    })

# Display results
risk_table = pd.DataFrame(risk_results)

display(risk_table.round(2))
```

```python

# Monte Carlo Historical Bootstrap Simulation

np.random.seed(42)

# Simulation settings
investment = 10000
n_simulations = 20000
time_horizon = 10

# Convert historical portfolio returns to an array
historical_returns = portfolio_returns.values

# Generate random daily returns for each simulation
simulated_returns = np.random.choice(
    historical_returns,
    size=(n_simulations, time_horizon),
    replace=True
)

# Calculate cumulative investment values
simulated_values = investment * np.prod(
    1 + simulated_returns,
    axis=1
)

# Calculate simulated losses
simulated_losses = investment - simulated_values

# Calculate simulated VaR and Expected Shortfall
monte_carlo_results = []

for confidence in [0.95, 0.99]:

    var = np.quantile(simulated_losses, confidence)

    es = simulated_losses[
        simulated_losses >= var
    ].mean()

    monte_carlo_results.append({
        'Confidence Level': f'{confidence:.0%}',
        'Monte Carlo VaR ($)': var,
        'Monte Carlo ES ($)': es
    })

monte_carlo_table = pd.DataFrame(monte_carlo_results)

print("10-Day Monte Carlo Risk Estimates:")
display(monte_carlo_table.round(2))

# Visualise simulated portfolio outcomes
plt.figure(figsize=(10, 5))

plt.hist(
    simulated_values,
    bins=60,
    edgecolor='black',
    alpha=0.75
)

plt.axvline(
    investment,
    linestyle='--',
    linewidth=2,
    label='Initial Investment ($10,000)'
)

plt.title('Monte Carlo Simulation: 10-Day Portfolio Values')
plt.xlabel('Portfolio Value ($)')
plt.ylabel('Number of Simulations')
plt.legend()
plt.grid(alpha=0.2)
plt.show()
```

```python

# Calculate historical 10-day compounded portfolio returns

time_horizon = 10
investment = 10000

rolling_10day_returns = (
    (1 + portfolio_returns)
    .rolling(window=time_horizon)
    .apply(np.prod, raw=True) - 1
).dropna()

# Convert returns into dollar losses
historical_10day_losses = -rolling_10day_returns * investment

# Calculate historical 10-day VaR and Expected Shortfall
comparison_results = []

for confidence in [0.95, 0.99]:
    var = historical_10day_losses.quantile(confidence)

    es = historical_10day_losses[
        historical_10day_losses >= var
    ].mean()

    comparison_results.append({
        'Confidence Level': f'{confidence:.0%}',
        'Historical 10-Day VaR ($)': var,
        'Historical 10-Day ES ($)': es
    })

historical_10day_table = pd.DataFrame(comparison_results)

print("Historical 10-Day Portfolio Risk Estimates:")
display(historical_10day_table.round(2))
```

```python

# Calculate cumulative portfolio growth
cumulative_portfolio = (1 + portfolio_returns).cumprod()

# Calculate total and annualised compounded returns
total_return = cumulative_portfolio.iloc[-1] - 1

years = len(portfolio_returns) / 252
cagr = cumulative_portfolio.iloc[-1] ** (1 / years) - 1

# Calculate maximum drawdown
running_peak = cumulative_portfolio.cummax()
drawdown = cumulative_portfolio / running_peak - 1
maximum_drawdown = drawdown.min()

print("PORTFOLIO PERFORMANCE SUMMARY")
print("--------------------------------")
print(f"Total Return: {total_return:.2%}")
print(f"Annualised Compounded Return (CAGR): {cagr:.2%}")
print(f"Maximum Drawdown: {maximum_drawdown:.2%}")

# Plot cumulative portfolio growth
plt.figure(figsize=(12, 6))
plt.plot(cumulative_portfolio, label='Equal-Weight Portfolio')
plt.title('Growth of the $1 Invested Equal-Weight Portfolio')
plt.xlabel('Date')
plt.ylabel('Portfolio Value ($)')
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

# Plot drawdown
plt.figure(figsize=(12, 5))
plt.fill_between(drawdown.index, drawdown, 0)
plt.title('Portfolio Historical Drawdown')
plt.xlabel('Date')
plt.ylabel('Drawdown')
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python

# OUT-OF-SAMPLE VaR BACKTESTING

split_index = int(len(portfolio_returns) * 0.70)

train_returns = portfolio_returns.iloc[:split_index]
test_returns = portfolio_returns.iloc[split_index:]

# Convert returns to losses
test_losses = -test_returns

backtest_results = []

for confidence in [0.95, 0.99]:

    # Historical VaR estimated from training data
    historical_var = -train_returns.quantile(1 - confidence)

    # Parametric VaR estimated from training data
    train_mean = train_returns.mean()
    train_std = train_returns.std()
    z_score = norm.ppf(1 - confidence)

    parametric_var = -(train_mean + z_score * train_std)

    # Count test-period losses exceeding each VaR threshold
    historical_exceptions = (test_losses > historical_var).sum()
    parametric_exceptions = (test_losses > parametric_var).sum()

    backtest_results.append({
        'Confidence Level': f'{confidence:.0%}',
        'Expected Exception Rate (%)': (1 - confidence) * 100,
        'Historical Exceptions': historical_exceptions,
        'Historical Exception Rate (%)':
            historical_exceptions / len(test_returns) * 100,
        'Parametric Exceptions': parametric_exceptions,
        'Parametric Exception Rate (%)':
            parametric_exceptions / len(test_returns) * 100
    })

backtest_table = pd.DataFrame(backtest_results)

print("OUT-OF-SAMPLE VaR BACKTEST RESULTS")
print("Training observations:", len(train_returns))
print("Testing observations:", len(test_returns))
display(backtest_table.round(2))
```

```python

from scipy.stats import binomtest

# Test observed exceptions against expected exception probabilities

test_observations = len(test_returns)

kupiec_results = []

for confidence in [0.95, 0.99]:

    expected_exception_probability = 1 - confidence

    historical_count = int(
        backtest_table.loc[
            backtest_table['Confidence Level'] == f'{confidence:.0%}',
            'Historical Exceptions'
        ].iloc[0]
    )

    parametric_count = int(
        backtest_table.loc[
            backtest_table['Confidence Level'] == f'{confidence:.0%}',
            'Parametric Exceptions'
        ].iloc[0]
    )

    historical_test = binomtest(
        historical_count,
        test_observations,
        expected_exception_probability
    )

    parametric_test = binomtest(
        parametric_count,
        test_observations,
        expected_exception_probability
    )

    kupiec_results.append({
        'Confidence Level': f'{confidence:.0%}',
        'Historical p-value': historical_test.pvalue,
        'Parametric p-value': parametric_test.pvalue
    })

kupiec_table = pd.DataFrame(kupiec_results)

print("VaR UNCONDITIONAL COVERAGE TEST")
display(kupiec_table.round(4))

print("\nInterpretation:")
print("A p-value below 0.05 indicates evidence that")
print("the observed exception frequency differs from")
print("the expected frequency at the 5% significance level.")
```
